1. Classes and Objects

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
    
    def display_info(self):
        print(f"Brand: {self.brand}\nModel: {self.model}\nYear: {self.year}")

toyota = Car("Toyota", "Camry", 2024)
tesla = Car("Tesla", "Model 3", 2025)

toyota.display_info()  # Output: Brand: Toyota, Model: Camry, Year: 2024
print('\n\n')
tesla.display_info()   # Output: Brand: Tesla, Model: Model 3, Year: 2025
```

Class methods

```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: ${amount}. New balance: ${self.balance}.")
        else:
            print("Deposit amount must be positive.")
    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew: ${amount}. New balance: ${self.balance}.")
        else:
            print("Withdrawal amount must be positive and less than or equal to the balance.")

    def display_balance(self):
        print(f"Account Holder: {self.account_holder}\nBalance: ${self.balance}")

account = BankAccount("Chirag", 10000)

account.deposit(5000)
account.withdraw(3000)
account.display_balance()
```

```python
import math
class ModelConfig:

    default_learning_rate = 0.001
    default_batch_size = 32


    def __init__(self, model_name, learning_rate, batch_size):
        self.model_name = model_name
        self.learning_rate = learning_rate
        self.batch_size = batch_size
    
    def show_config(self):
        print(f"Model Name: {self.model_name}\nLearning Rate: {self.learning_rate}\nBatch Size: {self.batch_size}")

    @classmethod
    def default_config(cls, model_name):
        return cls(model_name, cls.default_learning_rate, cls.default_batch_size)

    @staticmethod
    def is_valid_batch_size(batch_size):
        if batch_size <= 0:
            return False
        val = math.log2(batch_size)
        return val == int(val)
    
default_model = ModelConfig.default_config("ResNet50")
default_model.is_valid_batch_size(32)  # Output: True
```

3. Inheritance

```python
class Employee:
    company = "ABC.ai company"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def work(self):
        print(f"{self.name} is working at {self.company} with a salary of {self.salary}.")


class Developer(Employee):
    def __init__(self, name, salary,programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def code(self):
        print(f"{self.name} is coding in {self.programming_language}.")

chirag = Developer("Chirag", 100000, "Python")
chirag.work()  # Output: Chirag is working at ABC.ai company with a salary of 100000.
chirag.code()  # Output: Chirag is coding in Python.
```

4. Composition

```python
class CPU:
    def __init__(self):
        pass
    def process(self):
        print("CPU processing")
        return True

class RAM:
    def __init__(self):
        pass
    def load(self):
        print("RAM loading")
        return True

class Computer:
    def __init__(self):
        self.cpu = CPU()
        self.ram = RAM()

    def run(self):
        if self.cpu.process() and self.ram.load():
            print("Computer running")

    
my_computer = Computer()
my_computer.run()  # Output: CPU is processing data. RAM is loading data.
```

5. Encapsulation

```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self._balance = balance

    @property
    def balance(self):
        return self._balance
    
    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value
    
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited: ${amount}. New balance: ${self._balance}.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount
            print(f"Withdrew: ${amount}. New balance: ${self._balance}.")
        else:
            print("Invalid withdrawal amount or insufficient funds.")
```

6. Abstract Base Class

```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card.")

class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using UPI.")
```

```python
from abc import ABC, abstractmethod

class Model(ABC):

    def __init__(self, model_name):
        self._model_name = model_name

    @abstractmethod
    def predict(self, input_data):
        pass

    @property
    @abstractmethod
    def model_name(self):
        pass


class CNN(Model):
    def __init__(self):
        super().__init__("CNN")

    @property
    def model_name(self):
        return "CNN"

    @model_name.setter
    def model_name(self, value):
        self.__model_name = value

    def predict(self, data):
        print(f"Predicting using {self._model_name} on input data: {data}")
        return "Predicted Output"

class Transformer(Model):
    def __init__(self):
        super().__init__("Transformer")

    @property
    def model_name(self):
        return "Transformer"

    @model_name.setter
    def model_name(self, value):
        self.__model_name = value

    def predict(self, data):
        print(f"Predicting using {self._model_name} on input data: {data}")
        return "Predicted Output"
```

7. Dunder Methods and Property

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return math.pi * (self.radius ** 2)

c1 = Circle(5)
print(f"Area of Circle with radius {c1.radius}: {c1.area}")
```

```python
class Playlist:
    def __init__(self, songs):
        self._songs = songs
    
    def __len__(self):
        return len(self._songs)
    
    def __repr__(self):
        return f"Playlist with {len(self._songs)} songs: {', '.join(self._songs)}"
    
    def __getitem__(self, index):
        return self._songs[index]
         
playlist = Playlist(["Song A", "Song B", "Song C"])

print(playlist)
print(len(playlist))
print(playlist[1])
```

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return self.factor * value

double = Multiplier(2)
triple = Multiplier(3)

print(double(5))
print(triple(5))
```

```python
class Dataset:
    def __init__(self, data, labels):
        self._data = data
        self._labels = labels

    @property
    def size(self):
        return len(self._data)
    
    def __len__(self):
        return len(self._data)
    
    def __getitem__(self, index):
        return self._data[index], self._labels[index]
    
    def __repr__(self):
        return f"Dataset(size={len(self._data)})"

dataset = Dataset(
    data=["img1", "img2", "img3"],
    labels=[0, 1, 0]
)

print(dataset.size)
print(len(dataset))
print(dataset[1])
print(dataset)
```
