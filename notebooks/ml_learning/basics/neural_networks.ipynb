### Activation Functions

* **ReLU (Rectified Linear Unit)**
  $$ReLU(x) = \max(0, x)$$

* **Sigmoid**
  $$\sigma(x) = \frac{1}{1 + e^{-x}}$$

* **Tanh (Hyperbolic Tangent)**
  $$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

* **Softmax**
  $$\text{Softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{C} e^{z_j}}$$

* **GELU (Gaussian Error Linear Unit)**
  $$\text{GELU}(x) = x \cdot \Phi(x)$$

  *Framework Approximation:*
  $$\text{GELU}(x) \approx 0.5x \left(1 + \tanh\left(\sqrt{\frac{2}{\pi}} \left(x + 0.044715x^3\right)\right)\right)$$

```python
class MyNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(3,8)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(8,1)

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)

        return x


## nn.Sequential

model = nn.Sequential(
    nn.Linear(3,8),
    nn.ReLU(),
    nn.Linear(8,1)
)

x = torch.randn(5,3)
print(model(x))
```

### Loss Functions  

#### For Regression
* **MSE (Mean Squared Error / L2 Loss)**
  $$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$
* **MAE (Mean Absolute Error / L1 Loss)**  
  $$MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$

#### For Binary Classification
* **BCEWithLogitsLoss** (Combination of Sigmoid layer and BCE Loss)
  $$\begin{aligned}
  \text{Sigmoid: } & p = \sigma(x) = \frac{1}{1+e^{-x}} \\
  \text{BCE Loss: } & \text{Loss} = -\frac{1}{n}\sum_{i=1}^{n}[y_i \cdot \log(p_i) + (1 - y_i) \cdot \log(1 - p_i)]
  \end{aligned}$$

  *Where $y$ is the true label (0 or 1), $x$ is the raw logit, and $p$ is the predicted probability.*

#### For Multiclass Classification
* **CrossEntropyLoss** (Combination of Softmax layer and Categorical Cross Entropy)
  $$\text{Loss} = -\frac{1}{n}\sum_{i=1}^{n}\sum_{c=1}^{C} y_{ic} \cdot \log(p_{ic})$$

Gradient Descent

$$w_{new} = w - n.\frac{dL}{dw}$$

$w$: weight  
$n$: learning rate  
$\frac{dL}{dw}$: gradient

```python
## loss.backward(): calculates d(loss)/d(parameter) -> backpropogation

model = nn.Linear(3, 1)
print(model.weight.grad)
print(model.bias.grad)

x = torch.randn(5, 3)
target = torch.randn(5, 1)

prediction = model(x)

loss_fn = nn.MSELoss()

loss = loss_fn(prediction, target)

loss.backward()
```

```python
import torch
import torch.nn as nn

# Model
model = nn.Sequential(
    nn.Linear(3, 8),
    nn.ReLU(),
    nn.Linear(8, 3),
    nn.ReLU(),
    nn.Linear(3,1)
)

# Loss
loss_fn = nn.MSELoss()

# Optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

# Fake training data
x = torch.randn(100, 3, requires_grad=True)
y = torch.randn(100, 1, requires_grad=True)

for epoch in range(100):

    # 1. Clear old gradients
    optimizer.zero_grad()

    # 2. Forward pass
    prediction = model(x)

    # 3. Calculate loss
    loss = loss_fn(prediction, y)

    # 4. Backward pass
    loss.backward()

    # 5. Update parameters
    optimizer.step()

    if epoch % 10 == 0:
        print(epoch, loss.item())
```

Example of a classifier

```python
class Classifier(nn.Module):

    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(10, 32),
            nn.ReLU(),

            nn.Linear(32, 16),
            nn.ReLU(),

            nn.Linear(16, 3)
        )

    def forward(self, x):
        return self.network(x)
```

```python
model = Classifier()
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr = 0.001
)
```

```python
x = torch.randn(64, 10)

y = torch.randint(
    0,
    3,
    (64,)
)
```

```python
model.train()

for epoch in range(150):

    optimizer.zero_grad()

    logits = model(x)

    loss = loss_fn(logits, y)

    loss.backward()

    optimizer.step()

    print(
        f"Epoch {epoch}: loss = {loss.item():.4f}"
    )
```
