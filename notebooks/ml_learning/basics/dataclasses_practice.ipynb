```python
from dataclasses import dataclass, field
```

Examples of Dataclass

```python
@dataclass
class Account:
    name: str
    users: list = field(default_factory=list)

p1 = Account(name=123)
p2 = Account(name=456)
p1.users.append("chirag")
print(p1, p2)
```

```python
@dataclass
class Experiment:
    name: str
    metrics: dict = field(default_factory=dict)

e1 = Experiment("CNN")
e2 = Experiment("Transformer")

e1.metrics["accuracy"] = 0.95
print(e1, e2)
```

```python
from dataclasses import asdict
@dataclass
class ModelConfig:
    learning_rate: float
    batch_size: int
    optimizer: str = "adam"

@dataclass 
class OptimizerConfig:
    optimizer: str
    momentum: float = 0.9
    weight_decay: float = 0.0001

@dataclass
class TrainingConfig:
    model_config: ModelConfig
    optimizer_config: OptimizerConfig
    epochs: int = 10

training_config = TrainingConfig(
    model_config=ModelConfig(learning_rate=0.001, batch_size=32),
    optimizer_config=OptimizerConfig(optimizer="sgd")
)

(asdict(training_config))
```
