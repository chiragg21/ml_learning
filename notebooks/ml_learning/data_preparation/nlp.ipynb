```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)
```

```python
text = "I love coding."

tokens = tokenizer.tokenize(text)
tokens
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)
ids
```

```python
print(tokenizer.vocab_size, '\n', tokenizer.get_vocab())
```

```python
tokenizer("I love coding", add_special_tokens=True)
```

```python
texts = [
    "I love movies",
    "I really love this movie",
    "This was absolutely fantastic"
]

encoded = tokenizer(
    texts,
    padding=True
)

print(encoded["input_ids"])
print(encoded["attention_mask"])
```

**Important topics**
* Padding: adds padding tokens to match the sequence length.

* Truncation: Reduces sequence length

* Sequence length: length of vector containing token representation of sentence including special tokens.

* Special tokens: [CLS] -> start of sentence in BERT
                  [SEP] -> ending of sentence in BERT

<div style="border: 1px solid #241616; padding: 10px; border-radius: 5px; background-color: #6e3535;">
Choose sequence length appropriately, [example self attention ~ O(n^2)]
- suppose dataset has   
    50th percentile → 120 tokens  
    90th percentile → 250 tokens  
    95th percentile → 320 tokens  
    99th percentile → 700 tokens  
then choose max_length = 320
</div>

* Attention Mask: In earlier models [BERT], attention mask is a vector that is used so model focuses on actual text and not on padded tokens. But now [decoder/casual models such as GPT], masks can also be involved in controlling which positions can attend to which other positions.

**Test/Train/Val Split**  

split as Train - 90%, Val - 10% and use original test dataset.

In case of class imbalance,
* Strafied splitting
* Class-weighted loss: apply more weight on loss from minority classes
* Oversampling: add repeated samples of minority class, can cause overfitting
* Undersampling: reduce size of majority class, generally avoided as it is loss of avaible data
* Data Augmentation: Like paraphrasing, back translation and controlled word substitution

Text cleaning: lowercase, remove punctuation, remove stopwords, stemming, lemmatization

Data leakage: info from val/test influence training or preprocessing

<div style="border: 1px solid #241616; padding: 10px; border-radius: 5px; background-color: #6e3535;">  

Always do following:  
  
Split first   
   ↓  
Fit/learn preprocessing on TRAIN only  
   ↓  
Apply preprocessing to validation/test  
</div>
